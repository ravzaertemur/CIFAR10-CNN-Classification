```python
# Gerekli kütüphaneyi kuralım
!pip install -q gradio

import time
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, random_split
from torchvision import datasets, transforms
from sklearn.metrics import accuracy_score, precision_recall_fscore_support
import gradio as gr
from PIL import Image
import numpy as np

# 1. CİHAZ VE HİPERPARAMETRELER
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
BATCH_SIZE = 64
LEARNING_RATE = 0.001
EPOCHS = 15

classes = ['Uçak', 'Araba', 'Kuş', 'Kedi', 'Geyik', 'Köpek', 'Kurbağa', 'At', 'Gemi', 'Kamyon']

# 2. VERİ SETİ (CIFAR-10) VE TRAIN/VAL/TEST AYRIMI
transform = transforms.Compose([
    transforms.Resize((32, 32)),
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

full_train = datasets.CIFAR10(root='./data', train=True, download=True, transform=transform)
test_dataset = datasets.CIFAR10(root='./data', train=False, download=True, transform=transform)

# %80 Train (40.000), %20 Validation (10.000)
train_size = int(0.8 * len(full_train))
val_size = len(full_train) - train_size
train_dataset, val_dataset = random_split(full_train, [train_size, val_size])

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False)

# 3. CNN MODEL MİMARİSİ
class CustomCNN(nn.Module):
    def __init__(self):
        super(CustomCNN, self).__init__()
        self.features = nn.Sequential(
            # Blok 1: 32x32x3 -> 16x16x32
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),

            # Blok 2: 16x16x32 -> 8x8x64
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),

            # Blok 3: 8x8x64 -> 4x4x128
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128 * 4 * 4, 256),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(256, 10)
        )

    def forward(self, x):
        x = self.features(x)
        return self.classifier(x)

model = CustomCNN().to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)

# 4. EĞİTİM DÖNGÜSÜ
print("Eğitim başlıyor...")
start_train = time.time()

for epoch in range(EPOCHS):
    model.train()
    train_loss = 0.0
    for imgs, lbls in train_loader:
        imgs, lbls = imgs.to(device), lbls.to(device)
        optimizer.zero_grad()
        out = model(imgs)
        loss = criterion(out, lbls)
        loss.backward()
        optimizer.step()
        train_loss += loss.item()

    # Validation
    model.eval()
    val_correct, val_total = 0, 0
    with torch.no_grad():
        for imgs, lbls in val_loader:
            imgs, lbls = imgs.to(device), lbls.to(device)
            out = model(imgs)
            _, preds = torch.max(out, 1)
            val_total += lbls.size(0)
            val_correct += (preds == lbls).sum().item()

    print(f"Epoch [{epoch+1}/{EPOCHS}] - Train Loss: {train_loss/len(train_loader):.4f} | Val Acc: %{100*val_correct/val_total:.2f}")

train_duration = time.time() - start_train
print(f"\nToplam Eğitim Süresi: {train_duration:.2f} saniye")

# 5. TEST PERFORMANSI VE METRİKLER (Accuracy, Precision, Recall, F1, Inference Time)
model.eval()
all_preds, all_labels = [], []

start_inf = time.time()
with torch.no_grad():
    for imgs, lbls in test_loader:
        imgs = imgs.to(device)
        out = model(imgs)
        _, preds = torch.max(out, 1)
        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(lbls.numpy())
inf_duration = (time.time() - start_inf) / len(test_dataset) * 1000 # görüntü başına ms

acc = accuracy_score(all_labels, all_preds)
prec, rec, f1, _ = precision_recall_fscore_support(all_labels, all_preds, average='macro')

print("\n=== TEST SETİ PERFORMANS METRİKLERİ ===")
print(f"Accuracy : %{acc*100:.2f}")
print(f"Precision: {prec:.4f}")
print(f"Recall   : {rec:.4f}")
print(f"F1-Score : {f1:.4f}")
print(f"Ortalama Tahmin Süresi (Inference Time): {inf_duration:.4f} ms / görüntü")

# Modeli kaydet
torch.save(model.state_dict(), "cnn_cifar10.pth")

# 6. GRADIO WEB UYGULAMASI (DEPLOYMENT)
def tahmin_et(image):
    img = Image.fromarray(image.astype('uint8'), 'RGB')
    img_tensor = transform(img).unsqueeze(0).to(device)
    model.eval()
    with torch.no_grad():
        outputs = model(img_tensor)
        probs = torch.nn.functional.softmax(outputs[0], dim=0)
    return {classes[i]: float(probs[i]) for i in range(10)}

arayuz = gr.Interface(
    fn=tahmin_et,
    inputs=gr.Image(label="Test Edilecek Fotoğrafı Yükleyin"),
    outputs=gr.Label(num_top_classes=3, label="Model Tahminleri"),
    title="CIFAR-10 CNN Görüntü Sınıflandırma Uygulaması",
    description="Telefonunuzla çektiğiniz Araba, Kedi, Köpek, Kuş, Uçak vb. fotoğrafları yükleyerek modeli test edin."
)

arayuz.launch(share=True)
```
